```python
import ast

code = """
class CatDogDataset(Dataset):
    def __init__(self, csv_file, root_dir, transforms=None):
        self.cat_dog_frame = pd.read_csv(csv_file)
        self.root_dir = root_dir
        self.transforms = transforms

    def __len__(self):
        return len(self.cat_dog_frame)

    def __getitem__(self, idx):
        img_name = os.path.join(self.root_dir,
                                self.cat_dog_frame.iloc[idx, 0])

        image = Image.open(img_name)
        if self.transforms is not None:
            image_tensor = self.transforms(image)

        lbl = self.cat_dog_frame.iloc[idx, 1]
        return (image_tensor, lbl)
"""

tree = ast.parse(code)
lines = code.strip("\n").splitlines()

# Traverse AST
for node in tree.body:
    if isinstance(node, ast.ClassDef):
        print(f"\n📘 Class: {node.name}")
        class_start = node.lineno - 1
        class_end = node.end_lineno
        class_code = "\n".join(lines[class_start:class_end])
        print(f"\nFull Class:\n{class_code}")

        for method in node.body:
            if isinstance(method, ast.FunctionDef):
                method_start = method.lineno - 1
                method_end = method.end_lineno
                method_code = "\n".join(lines[method_start:method_end])
                print(f"\n🔧 Method `{method.name}`:\n{method_code}")
```

```python
import ast

code = """
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def mul(a, b):
    return a * b
"""

# Parse the AST
tree = ast.parse(code)

# Split the original code into lines (1-based index)
lines = code.strip("\n").splitlines()

for node in tree.body:
    if isinstance(node, ast.FunctionDef):
        start = node.lineno - 1  # lineno is 1-based
        end = node.end_lineno    # available in Python 3.8+
        function_code = "\n".join(lines[start:end])
        print(f"Function `{node.name}`:\n{function_code}\n")
```
